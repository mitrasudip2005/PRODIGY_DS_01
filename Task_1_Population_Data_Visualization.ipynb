```python
import pandas as pd

df = pd.read_csv("data.csv", skiprows=4)

print(df.head())
print(df.shape)
print(df.columns.tolist())
```

```python
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

print("\nColumn names:")
print(df.columns.tolist())

print("\nMissing values:")
print(df.isnull().sum())
```

```python
population_2024 = pd.to_numeric(df["2024"], errors="coerce")

population_2024 = population_2024.dropna()

print(population_2024.head())
print("Number of observations:", len(population_2024))
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))

plt.hist(population_2024, bins=20)

plt.title("Distribution of Population Across Countries in 2024")
plt.xlabel("Population (in million)")
plt.ylabel("Number of Countries")

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("data.csv", skiprows=4)

df["2024"] = pd.to_numeric(df["2024"], errors="coerce")

top_10 = df.nlargest(10, "2024")

print(top_10[["Country Name", "2024"]])
```

```python
plt.figure(figsize=(12, 6))

plt.bar(top_10["Country Name"], top_10["2024"])

plt.title("Top 10 Countries by Population in 2024")
plt.xlabel("Country")
plt.ylabel("Population")

plt.xticks(rotation=45, ha="right")

plt.tight_layout()
plt.show()
```

```python
exclude_words = [
    "World",
    "income",
    "IBRD",
    "IDA",
    "region",
    "dividend",
    "countries"
]

countries = df[
    ~df["Country Name"].str.contains(
        "|".join(exclude_words),
        case=False,
        na=False
    )
]

top_10 = countries.nlargest(10, "2024")
```

```python
plt.figure(figsize=(12, 6))

plt.bar(top_10["Country Name"], top_10["2024"])

plt.title("Top 10 Countries by Population in 2024")
plt.xlabel("Country")
plt.ylabel("Population")

plt.xticks(rotation=45, ha="right")

plt.tight_layout()
plt.show()
```
